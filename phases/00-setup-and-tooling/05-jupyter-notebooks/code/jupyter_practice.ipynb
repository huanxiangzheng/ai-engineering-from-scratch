# 随机数生成速度实验
比较 python 列表推导式和 Numpy 生成十万个随机数的耗时。

```python
import random
import numpy as np
```

```python
%timeit [random.random() for _ in range(100_000)]
```

```python
%timeit np.random.random(100_000)
```
